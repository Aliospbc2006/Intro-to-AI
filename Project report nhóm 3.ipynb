## 2. Đề xuất project (W3)

### Bài toán
**Tìm đường đi trong mạng Tokyo Metro** với tiêu chí cân bằng giữa "đi ngắn" và "ít đổi tuyến".

- **Đầu vào:** hai điểm trên bản đồ (xuất phát, đích) và tham số *Transfer Penalty* (mét).
- **Đầu ra:** chuỗi ga cần đi qua, tổng độ dài và số lần đổi tuyến.
- **Phạm vi:** chỉ Tokyo Metro, gồm 9 tuyến (G, M, H, T, C, Y, Z, N, F). Không gồm JR, Toei hay các tuyến tư nhân.
- **Biểu diễn:** mạng lưới là đồ thị, mỗi ga là một node, mỗi đoạn nối hai ga liền kề trên một tuyến là một edge có trọng số là độ dài (mét). Do đó, một ga chuyển tuyến chỉ là một node có các edge thuộc nhiều tuyến.
- **Sản phẩm:** ứng dụng web gồm giao diện bản đồ (Leaflet) và backend FastAPI + MongoDB.

### Phương pháp dự kiến
- **Thuật toán:** A\* trên không gian trạng thái `(node, line)` (đang ở ga nào, đang đi trên tuyến nào). Nhờ vậy thuật toán nhận biết được lúc nào hành khách đổi tuyến.
- **Hàm chi phí:** `g` là tổng độ dài các edge đã đi, cộng thêm *Transfer Penalty* mỗi khi `line` thay đổi.
- **Heuristic:** khoảng cách haversine từ ga hiện tại tới ga đích. Độ dài mỗi edge được làm tròn lên để luôn lớn hơn hoặc bằng haversine giữa hai đầu mút, nên heuristic không đánh giá quá cao và A\* cho đường đi tối ưu.
- **Dữ liệu:** danh sách ga và tọa độ từ MLIT N02-25 (CC BY 4.0); mã ga, tên tiếng Anh và màu tuyến đối chiếu với Wikidata (CC0). Dữ liệu được xử lý bằng script thành `nodes.json` và `edges.json`.
- **Kiến trúc:** Frontend (HTML + JavaScript + Leaflet) → FastAPI → MongoDB → đồ thị → A\*. Bản đồ nền lấy từ OpenStreetMap.
- **Kiểm chứng dự kiến:** viết script kiểm tra cấu trúc dữ liệu, đối chiếu A\* với Dijkstra (heuristic bằng 0) và thử qua giao diện.
